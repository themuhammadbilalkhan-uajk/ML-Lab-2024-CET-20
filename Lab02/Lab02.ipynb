Task 03: Seed 2026: one rand value, 25 randn values. Seed 42: roll 10,000 Ludo dice, count each face with np.unique, estimate P(six), count double sixes without a loop.

```python
np.random.seed(2026)
one_rand = np.random.rand()                 
many_randn = np.random.randn(25)            
print(f"T3a: one rand value in [0,1): {round(one_rand, 4)}")
print(f"T3b: 25 randn values (5x5 view):\n {np.round(many_randn.reshape(5, 5), 3)}")

np.random.seed(42)
dice = np.random.randint(1, 7, size=10000)  
print(f"T3c: first 15 rolls: {dice[:15]}")

faces, counts = np.unique(dice, return_counts=True)   
print(f"T3d: faces: {faces}")
print(f"T3d: counts (rolls): {counts}")

p_six = np.mean(dice == 6)                  
pairs = dice.reshape(5000, 2)               
double_sixes = np.sum(np.all(pairs == 6, axis=1))
print(f"T3e: P(six) = {p_six:.4f} (theory = {1/6:.4f})")
print(f"T3e: double sixes in 5,000 turns: {double_sixes} (theory ~ {5000/36:.0f})")
```

Task 06: Total; % share of each unit; most populous; units above 20 million; annual growth 2017–2023 in one line; fastest-growing unit. Interpret growth, and compare AJK.


```python

units = np.array(["Punjab", "Sindh", "KP", "Balochistan", "Islamabad", "AJK"])
pop17 = np.array([110.01, 47.85, 35.53, 12.34, 2.01, 4.05])
pop23 = np.array([127.69, 55.70, 40.85, 14.89, 2.36, 4.05])

def D(pairs):   
    return {str(k): (v.item() if hasattr(v, "item") else v) for k, v in pairs}

total = pop23.sum()
print(f"T6a: total population 2023 = {total:.2f} million")

share = pop23 / total * 100
print(f"T6b: share of total (%): {D(zip(units, np.round(share, 2)))}")

print(f"T6c: most populous unit: {units[np.argmax(pop23)]}")
print(f"T6d: units above 20 million: {units[pop23 > 20]}")

growth = ((pop23 / pop17) ** (1 / 6) - 1) * 100
print(f"T6e: annual growth 2017-2023 (% per year): {D(zip(units, np.round(growth, 2)))}")

print(f"T6f: fastest-growing unit: {units[np.argmax(growth)]} ({growth.max():.2f}% per year)")

nat = ((pop23.sum() / pop17.sum()) ** (1 / 6) - 1) * 100
doubling = 70 / growth.max()
print(f"T6 interpret: national growth = {nat:.2f}% per year.")
print(f"T6 interpret: {units[np.argmax(growth)]} would double in about {doubling:.0f} years at its current rate.")
print(f"T6 interpret: AJK grew {growth[-1]:.2f}% per year vs national {nat:.2f}% -> "
      f"AJK is {nat - growth[-1]:.2f} points below the national rate (in this sample AJK is essentially flat).")
```

Task 07: Shape; yearly mean per city; hottest month per city (argmax); 5-city monthly average; months above 35 °C; Lahore in °F; monsoon rain by slicing Jul–Sep; wettest monsoon city and monsoon share %. Interpret one surprise.

```python
cities = np.array(["Lahore", "Karachi", "Islamabad", "Quetta", "Peshawar"])
months = np.array(["Jan", "Feb", "Mar", "Apr", "May", "Jun",
                   "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"])

temp = np.array([   
    [19, 22, 28, 35, 40, 40, 36, 34, 34, 32, 26, 21],   
    [26, 28, 32, 34, 35, 34, 32, 31, 32, 34, 31, 28],   
    [17, 19, 24, 30, 36, 38, 35, 33, 33, 30, 24, 19],   
    [11, 13, 18, 24, 29, 34, 35, 34, 31, 25, 19, 13],   
    [18, 20, 25, 31, 37, 40, 38, 36, 35, 31, 25, 19],   
])
rain = np.array([   
    [12, 22, 34, 22, 21, 54, 214, 232, 82, 8, 3, 8],
    [8, 10, 9, 4, 1, 6, 40, 25, 13, 1, 2, 5],
    [56, 65, 76, 52, 32, 66, 260, 310, 120, 28, 12, 28],
    [38, 40, 33, 20, 8, 5, 18, 15, 5, 3, 7, 18],
    [33, 48, 63, 48, 23, 21, 86, 95, 43, 23, 20, 29],
])

def D(pairs):  
    return {str(k): (v.item() if hasattr(v, "item") else v) for k, v in pairs}

print(f"T7a: shape of temp (cities, months): {temp.shape}")

yearly_mean = temp.mean(axis=1)
print(f"T7b: yearly mean temperature (deg C): {D(zip(cities, np.round(yearly_mean, 1)))}")

hottest = months[np.argmax(temp, axis=1)]
print(f"T7c: hottest month per city: {D(zip(cities, hottest))}")

print(f"T7d: 5-city monthly average (deg C): {D(zip(months, np.round(temp.mean(axis=0), 1)))}")

above35 = (temp > 35).sum(axis=1)
print(f"T7e: months above 35 deg C: {D(zip(cities, above35))}")

lahore_f = temp[0] * 9 / 5 + 32
print(f"T7f: Lahore monthly temperature (deg F): {np.round(lahore_f, 1)}")

monsoon = rain[:, 6:9].sum(axis=1)
print(f"T7g: monsoon rain Jul-Sep (mm): {D(zip(cities, monsoon))}")
print(f"T7g: wettest monsoon city: {cities[np.argmax(monsoon)]} ({monsoon.max()} mm)")

share = monsoon / rain.sum(axis=1) * 100
print(f"T7h: monsoon share of annual rain (%): {D(zip(cities, np.round(share, 1)))}")

k = np.where(cities == "Karachi")[0][0]
print(f"T7 interpret: Karachi has the highest yearly mean ({yearly_mean.max():.1f} deg C) "
      f"but {above35[k]} months above 35 deg C, while cooler-on-average Peshawar has {above35[4]} "
      f"and Lahore {above35[0]}. A high yearly mean does not mean the most extreme heat: "
      f"Karachi is hot all year, the inland cities spike in summer.")
```

Task 08: Average per year; FY2024 months above 25 %; drop per month; month of biggest drop; label FY2025 months Low / Moderate / High with np.where. Interpret what happened to prices.



```python
fy = np.array(["FY2023", "FY2024", "FY2025"])
fmonths = np.array(["Jul", "Aug", "Sep", "Oct", "Nov", "Dec",
                    "Jan", "Feb", "Mar", "Apr", "May", "Jun"])

infl = np.array([
    [24.9, 27.3, 23.2, 26.6, 23.8, 24.5, 27.6, 31.5, 35.4, 36.4, 38.0, 29.4],
    [28.3, 27.4, 31.4, 26.9, 29.2, 29.7, 28.3, 23.1, 20.7, 17.3, 11.8, 12.6],
    [4.1, 9.6, 6.9, 7.2, 4.9, 4.1, 2.4, 1.5, 0.7, 0.3, 3.5, 3.2],
])

def D(pairs):
    return {str(k): (v.item() if hasattr(v, "item") else v) for k, v in pairs}

avg = infl.mean(axis=1)
print(f"T8a: average inflation per year (%): {D(zip(fy, np.round(avg, 2)))}")

fy24 = infl[1]
print(f"T8b: FY2024 months above 25 %: {fmonths[fy24 > 25]} -> {(fy24 > 25).sum()} months")

change = np.diff(infl, axis=1)
print(f"T8c: month-on-month change (percentage points; negative = drop):\n{np.round(change, 1)}")

biggest = np.argmin(change, axis=1)
print(f"T8d: month of biggest drop: {D(zip(fy, fmonths[biggest + 1]))} | size (pp): {np.round(change.min(axis=1), 1)}")

fy25 = infl[2]
label = np.where(fy25 < 5, "Low", np.where(fy25 < 10, "Moderate", "High"))
print(f"T8e: FY2025 labels (Low < 5 %, Moderate 5-10 %, High >= 10 %):\n{np.column_stack((fmonths, fy25, label))}")

print(f"T8 interpret: average inflation went {avg[0]:.1f}% -> {avg[1]:.1f}% -> {avg[2]:.1f}%. "
      f"Prices did NOT fall; they kept rising, but more and more slowly (disinflation), "
      f"so the price LEVEL is still much higher than at the start of FY2023.")
```

Task 09: +5 capped at 100 with np.clip; pass %; who passed only because of grace; z-scores.


```python
marks = np.array([34, 38, 35, 52, 67, 73, 88, 95, 98, 39,
                  41, 45, 36, 58, 62, 71, 80, 33, 99, 44])
rollno = np.arange(1, 21)
PASS = 40

new_marks = np.clip(marks + 5, 0, 100)
print(f"T9a: original marks: {marks}")
print(f"T9a: marks after +5 grace (cap 100): {new_marks}")

before = (marks >= PASS).mean() * 100
after = (new_marks >= PASS).mean() * 100
print(f"T9b: pass % before grace = {before:.1f} %, after grace = {after:.1f} %")

only_grace = (marks < PASS) & (new_marks >= PASS)
print(f"T9c: passed only because of grace (roll numbers): {rollno[only_grace]}")
print(f"T9c: their original marks: {marks[only_grace]}")

z = (new_marks - new_marks.mean()) / new_marks.std()
print(f"T9d: mean = {new_marks.mean():.2f}, std = {new_marks.std():.2f}")
print(f"T9d: z-scores (std devs from mean): {np.round(z, 2)}")
```

Task 10: Draw the "7" on an 8 × 8 canvas; flip; invert; crop; brighten with a cap; shape, dtype, white pixels.

```python
canvas = np.zeros((8, 8), dtype=np.uint8)
canvas[1, 1:7] = 255
canvas[[2, 3, 4, 5, 6], [6, 5, 5, 4, 4]] = 255

show = lambda a: (a > 0).astype(int)

print(f"T10a: the '7' (1 = white):\n{show(canvas)}")

flipped = np.fliplr(canvas)
print(f"T10b: flipped left-right:\n{show(flipped)}")

inverted = 255 - canvas
print(f"T10c: inverted (white <-> black):\n{show(inverted)}")

cropped = canvas[0:7, 1:8]
print(f"T10d: cropped 7x7 region, shape: {cropped.shape}\n{show(cropped)}")

bright = np.clip(canvas.astype(np.int16) + 60, 0, 255).astype(np.uint8)
print(f"T10e: brightened (+60, capped at 255); unique values: {np.unique(bright)}")

print(f"T10f: shape = {canvas.shape} | dtype = {canvas.dtype} | white pixels = {np.sum(canvas == 255)}")
```
