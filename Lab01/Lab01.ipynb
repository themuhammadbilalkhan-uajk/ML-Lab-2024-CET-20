09: Function: count how many times "dog" occurs in a string (case-insensitive).

```python
def dogCounter(text):
    return text.lower().count("dog")

print(f"Q9 P1: {dogCounter("The dog and the Dog and the dog.")} ")    
print(f"Q9 P2: {dogCounter("Dog run after the cat.")} ")    
print(f"Q9 P3: {dogCounter("Cats and cat")} ")        
```

Task 10

10: caught_speeding(speed, is_birthday): ≤ 60 "No Ticket", 61–80 "Small Ticket", ≥ 81 "Big Ticket". On a birthday, allow 5 mph more.

```python
def caught_speeding(speed, is_birthday):
    if is_birthday:
        speed = speed - 5

    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"

print(f"Q10 P1: {caught_speeding(60, False)} ")
print(f"Q10 P2: {caught_speeding(65, False)} ")
print(f"Q10 P3: {caught_speeding(90, False)} ")
        
```

Bonus Task 01

1: Make count_dog also count "dog," and "dog!" (punctuation).

```python
def PdogCounter(text):
    count = 0
    for word in text.lower().split():
        if word.strip(",.!?") == "dog":
            count += 1
    return count

print(f"BonusQ1 P1: {PdogCounter("The dog, a big dog! Dog? dogs")}")
print(f"BonusQ1 P2: {PdogCounter("dog, dog! dog.")} ") 
print(f"BonusQ1 P3: {PdogCounter("I love cats")} ") 
    
```
